```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df = data.copy()
```

```python
df["income_category"] = pd.cut(df['median_income'] , bins=[0,1.5,3.0,4.5,6.0 , np.inf],labels = [1,2,3,4,5])
```

```python
df.head()
```

```python
# data selection 
from sklearn.model_selection import train_test_split
train , test = train_test_split(df,test_size=0.2,random_state=42,stratify=df['income_category'])
```

```python
copy_of_train_data = train.copy() 
```

```python
# First We are drop median_house_value 
housing = copy_of_train_data.drop('median_house_value',axis=1)
housing_labels = copy_of_train_data['median_house_value'].copy
```

```python
# Handle Missing values 
from sklearn.impute import SimpleImputer 
imputer = SimpleImputer(strategy='median')
housing_number = housing.select_dtypes(include=[np.number])
imputer.fit(housing_number)
x_transformed = imputer.fit_transform(housing_number)
x_transformed
```

```python
# Fill The data
dataFrame = pd.DataFrame(x_transformed,columns=housing_number.columns,index=housing_number.index)
dataFrame
```

```python
# 1) Create imputer
imputer = SimpleImputer(strategy='median')

# 2) Select only numeric columns
housing_number = housing.select_dtypes(include=[np.number])

# 3) Fit + Transform (fill missing values)
housing_num_filled = pd.DataFrame(imputer.fit_transform(housing_number),
    columns=housing_number.columns,
    index=housing_number.index
)

# 4) Check missing values after filling
print("Before filling missing values:")
print(housing_number.isnull().sum())

print("\nAfter filling missing values:")
print(housing_num_filled.isnull().sum())
```
