```python
import pandas as pd
import numpy as np
from sklearn.model_selection import StratifiedShuffleSplit
```

```python
df = pd.read_csv("housing.csv")
```

```python
df['income_cat'] = pd.cut(df['median_income'], bins=[0, 1.5, 3.0, 4.5, 6.0, np.inf], labels=[1, 2, 3, 4, 5])
```

```python
from sklearn.model_selection import StratifiedShuffleSplit
split = StratifiedShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
for train_index, test_index in split.split(df, df['income_cat']):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]
    
```

```python
# Lets remove the income_cat column
for sett in (strat_train_set, strat_test_set):
    sett.drop("income_cat", axis=1, inplace=True)
```

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

data = pd.DataFrame({
    "City": ["Delhi", "Mumbai", "Delhi", "Jaipur"]
})

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(data[["City"]])

encoded_df = pd.DataFrame(encoded, columns=encoder.get_feature_names_out(["City"]))
print(encoded_df)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression

# Sample dataset
data = pd.DataFrame({
    "Size": ["1 BHK", "2 BHK", "3 BHK", "2 BHK", "1 BHK"],
    "City": ["Delhi", "Mumbai", "Delhi", "Jaipur", "Mumbai"],
    "Furnishing": ["Furnished", "Semi-Furnished", "Unfurnished", "Furnished", "Unfurnished"],
    "Price": [45, 75, 120, 80, 50]
})

X = data.drop("Price", axis=1)
y = data["Price"]

# Ordinal Encoding (Size)
size_order = [["1 BHK", "2 BHK", "3 BHK"]]

# Categorical columns for one-hot
nominal_cols = ["City", "Furnishing"]

preprocessor = ColumnTransformer(
    transformers=[
        ("size_encoder", OrdinalEncoder(categories=size_order), ["Size"]),
        ("onehot", OneHotEncoder(drop='first', sparse_output=False), nominal_cols)
    ],
    remainder="passthrough"
)

X_encoded = preprocessor.fit_transform(X)

print("Encoded Output:\n", X_encoded)
```
