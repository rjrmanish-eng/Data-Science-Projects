```python
import pandas as pd 
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
dataset = pd.read_excel('./iris.xlsx')
```

```python
dataset.head()
```

```python
# Select Data For Predications
x = dataset.iloc[:,:-1]
x
```

```python
# Select Data For Predications
y = dataset.loc[:,'class']
y
```

```python
# Set Modal Random Forest 
modal = RandomForestClassifier()
```

```python
# fit modal =>  
modal.fit(x.values,y.values)
```

```python
# predication Data 
predications = modal.predict([[5.5,3.5,1.3,0.2]])
```

```python
# predication Data 
predications1 =  modal.predict([[5.8,2.6,4,1.2]])
```

```python
# predication Data 
predications2 =  modal.predict([[7.4,2.8,6.1,1.9]])
```

```python
# Print Data Of Predication
print(f'The Class Of Iris is - {predications[0]}')
```

```python
# Print Data Of Predication
print(f'The Class Of Iris is - {predications1[0]}')
```

```python
# Print Data Of Predication
print(f'The Class Of Iris is - {predications2[0]}')
```
